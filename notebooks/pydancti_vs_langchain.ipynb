## Pydantic AI vs LangChain Speed Test

```python
from pydantic_ai import Agent, RunContext
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
from openai import OpenAI
from dotenv import load_dotenv
import nest_asyncio
from time import time
import os

load_dotenv('../.env')
nest_asyncio.apply() # for async issues in Jupyter stuff
```

Pydantic AI

```python
# Initialize the agent with OpenAI GPT model
agent = Agent(
    'openai:gpt-5-mini',
    system_prompt='You are a helpful assistant specialized in Python programming.',
)

async def pydantic_chatgpt():
# Run the agent asynchronously
    result = await agent.run('Explain how to use structured outputs in Pydantic AI in a short response')
    # print("ChatGPT Response:")
    # print(result.output)

# Run the async function
n_calls = 5
start = time()
for _ in range(n_calls):
    await pydantic_chatgpt()
end = time()

print(f"Execution time for {n_calls} calls={end-start}s")
```

LangChain

```python
model = ChatOpenAI(model="gpt-5-mini")

def langchain_chatgpt():
    messages = [
        SystemMessage(content='You are a helpful assistant specialized in Python programming.'),
        HumanMessage(content='Explain how to use structured outputs in Pydantic AI in a short response'),
    ]

    result = model.invoke(messages)
    # print(result.content)
    
n_calls = 5
start = time()
for _ in range(n_calls):
    langchain_chatgpt()
end = time()

print(f"Execution time for {n_calls} calls={end-start}s")
```

OpenAI API Straight

```python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY")) 

def openai_chatgpt():
    response = client.chat.completions.create(
        model="gpt-5-mini",
        messages=[
            {"role": "system", "content": 'You are a helpful assistant specialized in Python programming.'},
            {"role": "user", "content": 'Explain how to use structured outputs in Pydantic AI in a short response'}
        ]
    )

n_calls = 5
start = time()
for _ in range(n_calls):
    openai_chatgpt()
end = time()

print(f"Execution time for {n_calls} calls={end-start}s")
```

Conclusion: There appears to be little difference between the calling methods.
